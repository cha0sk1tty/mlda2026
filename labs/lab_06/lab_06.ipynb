```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

## Генерируем данные

```python
# генерируем 200 точек с нелинейной зависимостью на основе синусоид + случайный шум
X = np.linspace(0, 6, 200)[:, np.newaxis]
y = np.sin(X).ravel() + np.sin(6 * X).ravel() + np.random.RandomState(1).normal(0, 0.1, X.shape[0]) + 3
```

```python
y[:5]
```

```python
plt.figure(figsize=(9, 6))
plt.scatter(X[:, 0], y)

plt.xlabel('X', size=18)
plt.ylabel('y', size=18)
plt.grid()
plt.show()
```

## Train / Test Split

```python
from sklearn.model_selection import train_test_split

# разделяем данные на тестовую и обучающую выборку в соотношении 50/50
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

## Обучаем регрессоры

Будем использовать три модели:
* kNN
* Древо решений
* Линейная регрессия

Мы будем использовать scikit-learn реализацию этих регрессоров. Их описания: [kNN](https://scikit-learn.org/stable/modules/generated/sklearn.neighbors.KNeighborsRegressor.html), [Decision Tree](https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeRegressor.html), [Logistic Regression](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html).

Давайте просто импортируем их и обучим.

```python
# Import kNN regressor
from sklearn.neighbors import KNeighborsRegressor

# Create object of the regressor's class
knn_reg = KNeighborsRegressor(n_neighbors=10)

# Fit the regressor
knn_reg.fit(X_train, y_train)
```

```python
# Import Decision Tree regressor
from sklearn.tree import DecisionTreeRegressor

# выбираем лучшее разделение, уменьшающее сумму квадратов ошибок 
dt_reg = DecisionTreeRegressor(criterion='squared_error', splitter='best', max_depth=4, 
                            min_samples_split=2, min_samples_leaf=1)

# обучение модели на входных данных и правильных ответах
dt_reg.fit(X_train, y_train)
```

```python
# Import Linear Regression regressor
from sklearn.linear_model import LinearRegression

# Create object of the regressor's class
linreg = LinearRegression()

# Fit the regressor
linreg.fit(X_train, y_train)
```

## Прогнозы

```python
# kNN - предсказания модели для новых данных
y_test_knn_reg = knn_reg.predict(X_test)

# DT
y_test_dt_reg = dt_reg.predict(X_test)

# LinReg
y_test_linreg = linreg.predict(X_test)
```

```python
#графики предсказаний моделей для сравнения с реальными значениями
plt.figure(figsize=(15, 6))
plt.scatter(X_test[:, 0], y_test, color='0', label='Truth')

sortd_inds = np.argsort(X_test[:, 0])
plt.plot(X_test[sortd_inds, 0], y_test_knn_reg[sortd_inds], linewidth=3, color='b', label='kNN') # более плавный
plt.plot(X_test[sortd_inds, 0], y_test_dt_reg[sortd_inds], linewidth=3, color='r', label='DT') # ступенчатый
plt.plot(X_test[sortd_inds, 0], y_test_linreg[sortd_inds], linewidth=3, color='g', label='LinReg') # ограничен линейной зависимостью

plt.xlabel('X', size=18)
plt.ylabel('y', size=18)
plt.legend(loc='best', fontsize=14)
plt.grid()
plt.show()
```

## Метрики качества

**1. (R)MSE ((Root) Mean Squared Error)** (штрафует больше всего, возводит ошибки в квадрат)

$$ L(\hat{y}, y) = \frac{1}{N}\sum\limits_{n=1}^N (y_n - \hat{y}_n)^2$$

**2. MAE (Mean Absolute Error)** (абсолютные значения, меньше штрафует)

$$ L(\hat{y}, y) = \frac{1}{N}\sum\limits_{n=1}^N |y_n - \hat{y}_n|$$

**3. RSE (Relative Squared Error)** (прогноз относительно ср.значения, возводит ошибки в квадрат)

$$ L(\hat{y}, y) = \sqrt\frac{\sum\limits_{n=1}^N (y_n - \hat{y}_n)^2}{\sum\limits_{n=1}^N (y_n - \bar{y})^2}$$

**4. RAE (Relative Absolute Error)** (прогноз относительно ср.значения, абсолютные значения, меньше штрафует)

$$ L(\hat{y}, y) = \frac{\sum\limits_{n=1}^N |y_n - \hat{y}_n|}{\sum\limits_{n=1}^N |y_n - \bar{y}|}$$

**5. MAPE (Mean Absolute Persentage Error)** (измеряет ошибку в процентах от предсказуемого значения, если ошибка большая, а значение маленькое, то процент большой)

$$ L(\hat{y}, y) = \frac{100}{N} \sum\limits_{n=1}^N\left|\frac{ y_n - \hat{y}_n}{y_n}\right|$$


**6. RMSLE (Root Mean Squared Logarithmic Error)** (сглаживает большие абсолютные значения, полезна, когда важно учитывать относительные различия между положительными значениями)

$$ L(\hat{y}, y) = \sqrt{\frac{1}{N}\sum\limits_{n=1}^N(\log(y_n + 1) - \log(\hat{y}_n + 1))^2}$$

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error, mean_squared_log_error

def regression_quality_metrics_report(y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    rse = np.sqrt(np.sum((y_true - y_pred) ** 2) / np.sum((y_true - np.mean(y_true)) ** 2))
    rae = np.sum(np.abs(y_true - y_pred)) / np.sum(np.abs(y_true - np.mean(y_true)))
    mape = mean_absolute_percentage_error(y_true, y_pred) * 100
    rmsle = np.sqrt(mean_squared_log_error(y_true, y_pred))

    return [rmse, mae, rse, rae, mape, rmsle]
```

```python
metrics_report = pd.DataFrame(columns=['RMSE', 'MAE', 'RSE', 'RAE', 'MAPE', 'RMSLE'])

metrics_report.loc['kNN', :] = regression_quality_metrics_report(y_test, y_test_knn_reg)
metrics_report.loc['DT', :] = regression_quality_metrics_report(y_test, y_test_dt_reg)
metrics_report.loc['LinReg', :] = regression_quality_metrics_report(y_test, y_test_linreg)

metrics_report
```

**Вопросы**:
- Какой регрессор лучше? Дерево решений и метод ближайших соседей
- Как вы можете улучшить качество моделей? Подобрать гиперпараметры, увеличить объем обучающей выборки. Качество соседей для kNN, глубину дерева для древо решений
- Как вы можете объяснить поведение регрессоров? kNN хорошо повторяет плавные изменения, дерево решений делит данные на участки, так как ступенчатый вид, линейная ограничена линейной зависимостью.
- Что будет, если изменить их гиперпараметры? kNN: уменьшение количества соседей делает модель чувствительнее к шуму, увеличение — сглаживает предсказания. Дерево решений: увеличение глубины позволяет точнее описывать данные, но повышает риск переобучения. Уменьшение глубины может привести к недообучению. Линейная регрессия: изменение стандартных параметров не позволит описать нелинейную зависимость. Для улучшения можно добавить полиномиальные (в степень возвести) признаки.
