
## Ответы на вопросы для самопроверки

### 1. Когда определитель $\Delta=n\sum x_i^2-(\sum x_i)^2$ равен нулю?

Он равен нулю тогда и только тогда, когда все $x_i$ одинаковы. Экспериментальные точки располагаются на вертикальной прямой. Линейная модель $y=ax+b$ не имеет единственных коэффициентов: одни и те же предсказания при данном $x$ могут получаться при разных $a,b$. Приближение по МНК возможно, но параметры не идентифицируются единственным образом.

### 2. Почему матрица Гессе положительно определена?

Для линейной функции

$$H=2\begin{pmatrix}\sum x_i^2&\sum x_i\\\sum x_i&n\end{pmatrix}.$$

Если не все $x_i$ равны, её ведущие главные миноры положительны по критерию Сильвестра. Значит, функция потерь строго выпуклая, а единственная стационарная точка является единственным глобальным минимумом. При одинаковых $x_i$ положительной определённости нет.

### 3. Может ли измениться рейтинг четырёх функций при выборе другого критерия?

Да. В образце из методички, если пересчитать ошибки для приведённых там округлённых функций, при переходе от суммы квадратов абсолютных отклонений к сумме квадратов логарифмических отклонений линейная функция опускается со второго на четвёртое место, а показательная поднимается с четвёртого на второе. Квадратичная остаётся первой. Это происходит потому, что логарифмическая ошибка приблизительно соответствует относительной, а не абсолютной.

### 4. Как выявить ошибки в формулах Крамера без полного ручного пересчёта?

Можно подставить найденные коэффициенты в исходные нормальные уравнения и проверить, близки ли левые части к правым. Например, в Python: `np.allclose(A @ coefficients, rhs)`. Ручной счёт помогает разобраться в методе, но для проверки расчётов надёжнее использовать компьютер.

### 5. Всегда ли меньшая ошибка квадратичной функции означает, что она лучше?

Нет. Квадратичная модель имеет больше параметров, поэтому на обучающих точках её минимальная сумма квадратов не превышает линейную. Однако это не гарантирует точности для новых данных. Улучшение можно оценить перекрёстной проверкой, например Leave-One-Out, или с учётом сложности модели.

### 6. Достаточно ли одного графика для выбора модели?

Нет. По нему сложно увидеть точные остатки, сумму их квадратов и устойчивость прогноза. Требуются количественные показатели ошибок и, по возможности, проверка на новых данных.

### 7. Почему нельзя логарифмировать нулевые и отрицательные значения $y$?

В действительных числах $\ln y$ определён лишь при $y>0$. При $y\leq0$ следует выбрать другую модель или использовать нелинейную минимизацию ошибок без логарифмирования. Произвольное смещение значений $y$ меняет постановку задачи.

### 8. Как изменяется ошибка из-за округления коэффициентов?

Для варианта 18 значения до и после округления приведены в таблице сравнения. Изменение составляет: линейная $+0{,}0007$, степенная $+0{,}0110$, показательная $-2{,}9222$, квадратичная $+0{,}1192$. Для проверки достаточно пересчитать функцию потерь при округлённых коэффициентах — решать систему заново не нужно.

### 9. Как связаны критерий Сильвестра и сингулярность матрицы нормальных уравнений? Почему используют `np.linalg.lstsq`?

Для модели $A\theta\approx y$ матрица Гессе равна $2A^TA$. Если столбцы $A$ линейно независимы, $A^TA$ положительно определена, и выполняется критерий Сильвестра. При линейной зависимости столбцов она сингулярна. `np.linalg.lstsq` решает задачу непосредственно, используя численно более устойчивые методы, чем формулы Крамера или явное обращение $A^TA$.

### 10. Можно ли получить меньшую ошибку у линейной функции, чем у квадратичной?

**При точном решении МНК по одним и тем же точкам и одинаковой сумме квадратов — нет.** Линейная модель является частным случаем квадратичной при коэффициенте $a=0$, поэтому $L_{\text{квадр}}\leq L_{\text{лин}}$. Например, для $x=1,2,3,4,5,6$ и $y=3,5,7,9,11,13$ обе модели могут иметь нулевую ошибку. Линейная модель может оказаться предпочтительнее по простоте или качеству предсказания новых точек, а также иметь меньшую **пересчитанную** ошибку после округления коэффициентов.
